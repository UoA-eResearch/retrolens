# Drivers 06 — Rivers & precipitation (REC2 v5)
River mouths = terminal REC2 segments within 1.5 km of the MHW coastline, with
a runoff-based discharge proxy validated against Clutha/Waikato magnitudes.
Per transect: distance to nearest (big) mouth, exponential-decay sediment
supply index (L = 10/20/50 km), coastal rainfall (`seg_rain` of the nearest
segment) and storm-rain days. No gauged national discharge dataset exists —
REC2 modelled runoff is the proxy; noted as a limitation.
