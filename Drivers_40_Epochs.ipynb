# Drivers 40 — Epoch rates (pre/post-1990)
Recomputes WLR/LRR/EPR per epoch from the embedded per-transect shoreline time
series with a closed-form WLS/OLS engine (`drivers/epochs.py`) that reproduces
statsmodels to <1e-6 and the published full-period rates (LRR/EPR 100% within
0.01 m/yr; the known WLR/NSM caveats are documented in
`qa/epoch_validation.txt`). Splits at 1985/1990/1995 for sensitivity —
the 1990s observation trough makes split placement consequential.

```python
import pandas as pd
from drivers import common

path = common.DRIVER_DATA / "epoch_rates.parquet"
if not path.exists():
    from drivers.epochs import build
    ep = build()
else:
    ep = pd.read_parquet(path)
ep.groupby("split_year")[["both_regression","both_epr"]].sum()
```

## Coverage
~108k transects (47%) have ≥3 shorelines in **both** epochs at the 1990 split —
the epoch analysis works at transect level.

```python
e90 = ep[(ep.split_year==1990) & ep.both_regression]
e90[["WLR_pre","WLR_post","delta_WLR","n_obs_pre","n_obs_post"]].describe()
```
